# Guideline on the challenges and scoring

There are 3 challenges, and each one contributes a certain number of points, which are distributed across its exercises. Most exercises use binary grading, i.e. if you pass you earn the full points for that exercise. However, each challenge also includes a final exercise, which is more complex and serves as the culmination of the lab. This exercise is graded in a continuous way, meaning your score depends on the quality of your solution and can range from 0 up to the exercise's maximum. The score for each exercise is indicated in each challenge's notebook.

After the challenge deadline, your team will receive a final score based on the total points accumulated by running the notebooks and passing the graders. There is no limit on the number of attempts to try to pass an exercise. Once you pass an exercise, its grade is saved on our server, so you don't need to worry if you accidentally delete your solution later.

Based on your total score, teams are classified into the following categories:

### Challenge 1: QMOO (100 points)

**The maximum grade for this exercise is 100 points**

*Summary:*

In this challenge, we will tackle a multi-objective optimization (MOO) problem, reformulated as a weighted Max-Cut problem that can be solved by the Quantum Approximate Optimization Algorithm (QAOA). In particular your task is to to prepare a suitable ansatz for solving the multi-objective Max-cut problem, first for a problem with 42 qubits and then you will be asked to scale it to 80 qubits.

### Challenge 2: Sample-based Krylov Quantum Diagonalization (100 points)

**The maximum grade for this exercise is 100 points**

*Summary:*

In this challenge, we implement a Sample-based Krylov Quantum Diagonalization (SKQD) workflow to approximate ground states of the transverse-field Ising model and single-impurity Anderson model, executing on IBM Quantum hardware to achieve results comparable to classical exact methods.

### Challenge 3: Hadron-Schwinger (100 points)

**The maximum grade for this exercise is 100 points**

*Summary:*

In this challenge, we will simulate the propagation of a hadron wavepacket in the Schwinger model on an IBM quantum computer, prepared on a lattice with $L=34$ spatial sites (68 qubits). We will start with initial states already prepared by SC-ADAPT-VQE and perform the time evolution of the Hamiltonian using the Suzuki-Trotter algorithm to evolve the initial states forward in time. The circuits will first be executed on a simulator and later on an IBM quantum computer.

# Installation guide (Mac and Linux)

### Step 1: Create virtual environment.

Please use Python 3.11-3.13. If you do not have one of these versions of Python installed on your machine, please follow the instructions [here](https://www.python.org/downloads/). Then, run the following code in your open terminal to create a new environment:

```python3 -m venv r2p-venv```

Then activate the environment:

```source r2p-venv/bin/activate```
